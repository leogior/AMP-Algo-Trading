```python
import numpy as np
import pandas as pd
from tqdm import tqdm


from backtesterClass.orderClass import orders
from backtesterClass.orderBookClass import OBData
from backtesterClass.analysisClass import analysisClass


from strats.movingAverageStrat import movingAverageStrat
from strats.rsiStrat import rsiStrat
from strats.momentumStrat import momentumStrat
from strats.momentumOnlineLearn import momentumOnlineLearnStrat


from utils.utils import global_perf, peformance_metrics

from cProfile import Profile
from pstats import SortKey, Stats
import gc


# Improve computationnal performance of the backtester - increases garbage collector threshold

allocs, gen1, gen2 = gc.get_threshold()
allocs = 5000
gen1 = gen1*2
gen2=gen2*2
gc.set_threshold(allocs, gen1,gen2)
```

##

## Load and Initialize data

```python
db_path = r"data/all_ohlcv_data.csv"
dataClass = OBData(db_path)
```

## Moving Average Strategies

```python
movingAverageTrader_5_20 = movingAverageStrat("movingAverageTrader_5_20", short_window=5, long_window=20)
movingAverageTrader_15_45 = movingAverageStrat("movingAverageTrader_15_45", short_window= 15, long_window=45)
movingAverageTrader_10_50 = movingAverageStrat("movingAverageTrader_10_50", short_window= 10, long_window=50)
movingAverageTrader_20_100 = movingAverageStrat("movingAverageTrader_20_100", short_window= 20, long_window=100)
movingAverageTrader_50_200 = movingAverageStrat("movingAverageTrader_50_200", short_window= 50, long_window=200)
```

## Backtesting 

```python
for _ in tqdm(range(len(dataClass.OBData_))):
    orderClass = orders()
    movingAverageTrader_5_20.strategy(orderClass)
    movingAverageTrader_15_45.strategy(orderClass)
    movingAverageTrader_10_50.strategy(orderClass)
    movingAverageTrader_20_100.strategy(orderClass)
    movingAverageTrader_50_200.strategy(orderClass)
    OBData.step +=1 
```

## Single Stock and Global Analysis - Moving Average

```python
analysisMA = analysisClass(movingAverageTrader_5_20, path=r"Results", dashboardName="executionMovingAverageStrat", dbName="dbMovingAverageStrat")
analysisMA.create_dashboard("AAPL", show=True)
```

```python
# Strategy instances
strategies = {
    "5-20": movingAverageTrader_5_20,
    "15-45": movingAverageTrader_15_45,
    "10-50": movingAverageTrader_10_50,
    "20-100": movingAverageTrader_20_100,
    "50-200": movingAverageTrader_50_200,
}

global_perf(strategies, "Moving Average")

peformance_metrics(strategies,verbose=True)
```
